```python
from pathlib import Path
import numpy as np
import pandas as pd

ROOT = Path.cwd() if (Path.cwd() / "data").exists() else Path.cwd().parent
FOLDER = ROOT / "data" / "numpy"
names = (FOLDER / "feature_names.txt").read_text().split()

X_dev, y_dev = np.load(FOLDER / "X_dev.npy"), np.load(FOLDER / "y_dev.npy")
X_test, y_test = np.load(FOLDER / "X_test.npy"), np.load(FOLDER / "y_test.npy")
print("training:", X_dev.shape, " test:", X_test.shape)
```

```python
# All 253,680 people of 2015 in one table (one row = one person)
parts = []
for label, X, y in (("training", X_dev, y_dev), ("test", X_test, y_test)):
    d = pd.DataFrame(X, columns=names).astype(int)
    d.insert(0, "part", label)
    d["Diabetes"] = y
    parts.append(d)
df = pd.concat(parts, ignore_index=True)
df
```

```python
# Summary of every column: count, average, smallest and largest value
df.describe().round(2).T
```

```python
# How many people have diabetes in each part
df.groupby("part")["Diabetes"].agg(["count", "sum", "mean"]).rename(columns={"count": "people", "sum": "with diabetes", "mean": "share"})
```

```python
# Later survey years
for year in (2021, 2023):
    d = np.load(FOLDER / f"brfss{year}.npz")
    print(year, d["X"].shape, f"diabetes {d['y'].mean():.1%}")
```
